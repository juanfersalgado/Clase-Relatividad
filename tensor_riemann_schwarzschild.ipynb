```python
import einsteinpy.symbolic as es
import sympy as sp
import numpy as np
from IPython.display import display
```

```python
T, x, y, z = sp.symbols('T x y z')

# Minkowski en coordenadas cartesianas
gmat = np.diag([1, -1, -1, -1])
g = es.MetricTensor(gmat, [T, x, y, z])
display(g.tensor())
```

```python
# Símbolos de Christoffel
Gamma = es.ChristoffelSymbols.from_metric(g)
display(Gamma.tensor())
```

```python
T, r, teta, z = sp.symbols('T r teta z')

# Minkowski en coordenadas cilíndricas ds^2 = dT^2 - dr^2 - r^2 dteta^2 - dz^2
gmat = np.diag([1, -1, -r**2, -1])
g = es.MetricTensor(gmat, [T, r, teta, z])
Gama = es.ChristoffelSymbols.from_metric(g)
display(Gama.tensor())
```

```python
Riemann = es.RiemannCurvatureTensor.from_metric(g)
display(Riemann.tensor())
```

Métrica con curvatura

```python
# Coordenadas esféricas: r, teta, phi
# Elemento de línea espacial: dl^2= dr^2 + r^2 dteta^2 + r^2 sin^2(teta) dphi^2
# Métrica de Newton: ds^2 = (1 - 2GM/(rc^2))dT^2 - dr^2 - r^2 dteta^2 - r^2 sin^2(teta) dphi^2
T, r, teta, fi, G, M, c = sp.symbols('T r theta phi G M c')

gmat = np.diag([1 - 2*G*M/(r*c**2), -1, -r**2, -r**2*sp.sin(teta)**2])
g = es.MetricTensor(gmat, [T, r, teta, fi])
display(g.tensor())
```

```python
Riemann = es.RiemannCurvatureTensor.from_metric(g)
Rtensor = Riemann.tensor()
display(Rtensor)
```

Obtener el tensor completamente covariante

```python
gRiemann_gonorrea = sp.tensorproduct(g.tensor(), Riemann.tensor())

Riemann_cov = sp.tensorcontraction(gRiemann_gonorrea, (1, 2))
display(Riemann_cov)
```

Manera más rápida

```python
Riemann_cov = Riemann.change_config('llll')
Rcov = Riemann_cov.tensor()
display(Rcov)
```

```python
display(Rcov[1,0,0,1]), display(Rcov[0,1,1,0])
```

Tensor de Ricci

Con sympy

```python
ginv = sp.Matrix(g.tensor()).inv()
ginv
```

```python
g.config
```

```python
ginv = g.change_config('uu')
ginv.tensor()
```

```python
gRiemann_ricci = sp.tensorproduct(ginv.tensor(), Riemann_cov.tensor())

Ricci_tensor = sp.simplify(sp.tensorcontraction(gRiemann_ricci, (0, 2), (1, 5)))
display(Ricci_tensor)
```

Con Einsteinpy

```python
Ricci = es.RicciTensor.from_metric(g)
display(Ricci.tensor())
```

Traza del Tensor de Ricci

```python
# R_mu_nu, R^mu_nu = g^{mu nu} R_mu_nu ---->
gRicci = sp.tensorproduct(ginv.tensor(), Ricci.tensor())

R = sp.simplify(sp.tensorcontraction(gRicci, (0, 2), (1, 3)))
R # Curvatura de Riemann escalar
```

```python
R = es.RicciScalar.from_metric(g)
display(R.tensor())
```

Tensor de Ricci de una métrica radialmente simétrica

```python
# Coordenadas esféricas: r, teta, phi
# Elemento de línea espacial: dl^2= dr^2 + r^2 dteta^2 + r^2 sin^2(teta) dphi^2
# Métrica de Newton: ds^2 = (1 - 2GM/(rc^2))dT^2 - dr^2 - r^2 dteta^2 - r^2 sin^2(teta) dphi^2
T, r, teta, fi, G, M, c = sp.symbols('T r theta phi G M c')

A = sp.Function('A')
B = sp.Function('B')

gmat = np.diag([A(r), -B(r), -r**2, -r**2*sp.sin(teta)**2])
g = es.MetricTensor(gmat, [T, r, teta, fi])
display(g.tensor())
```

Siempre usar el signo '-' para compensar por la convención de signos del curso

```python
Ricci = es.RicciTensor.from_metric(g)
display(-Ricci.tensor())
```
